```
# 1. Install required packages
!pip install -q pandas

import pandas as pd
from typing import Optional, List
from IPython.display import HTML, display

# ==========================================
# PHASE 1: DATA ACQUISITION & SIMULATION
# ==========================================
def get_mock_job_database() -> List[dict]:
    """Simulates a raw database payload with 10 Chemical Engineering & Chemistry roles."""
    return [
        {"id": 1, "title": "Senior Process Engineer", "company": "BASF", "location": "Stuttgart", "type": "Full-Time", "salary": 92000, "category": "engineering"},
        {"id": 2, "title": "Analytical Chemist", "company": "Merck Group", "location": "Darmstadt", "type": "Full-Time", "salary": 74000, "category": "chemistry"},
        {"id": 3, "title": "R&D Materials Scientist", "company": "Evonik", "location": "Stuttgart", "type": "Full-Time", "salary": 85000, "category": "chemistry"},
        {"id": 4, "title": "Biochemical Plant Consultant", "company": "ChemConsult", "location": "Remote", "type": "Contract", "salary": 110000, "category": "engineering"},
        {"id": 5, "title": "Polymer Synthesis Specialist", "company": "Covestro", "location": "Leverkusen", "type": "Full-Time", "salary": 78000, "category": "chemistry"},
        {"id": 6, "title": "Environmental Safety Engineer", "company": "EcoChem Solutions", "location": "Stuttgart", "type": "Full-Time", "salary": 81000, "category": "engineering"},
        {"id": 7, "title": "Pharmaceutical Quality Analyst", "company": "Bayer", "location": "Frankfurt", "type": "Part-Time", "salary": 65000, "category": "chemistry"},
        {"id": 8, "title": "Catalysis Research Lead", "company": "Max Planck Institute", "location": "Stuttgart", "type": "Full-Time", "salary": 88000, "category": "chemistry"},
        {"id": 9, "title": "Renewable Energy Chemical Engineer", "company": "H2-Energy Tech", "location": "Munich", "type": "Full-Time", "salary": 96000, "category": "engineering"},
        {"id": 10, "title": "Process Safety Specialist", "company": "Linde plc", "location": "Stuttgart", "type": "Contract", "salary": 102000, "category": "engineering"}
    ]

def fetch_jobs(keyword: str, location: Optional[str] = None) -> pd.DataFrame:
    """Acquires chemical/chemistry listings and structures them into a Pandas DataFrame."""
    raw_data = get_mock_job_database()
    df = pd.DataFrame(raw_data)

    # Filter by keyword across title or category fields (case-insensitive)
    if keyword:
        df = df[df['title'].str.contains(keyword, case=False) | df['category'].str.contains(keyword, case=False)]

    # Filter by location if specified
    if location:
        df = df[df['location'].str.lower() == location.lower()]

    return df

# ==========================================
# PHASE 2: DATA PROCESSING
# ==========================================
def process_job_metrics(df: pd.DataFrame):
    """Computes operational KPIs and processes columns explicitly for the layout view."""
    if df.empty:
        return 0, "€0", df

    total_listings = len(df)
    avg_salary = f"€{round(df['salary'].mean(), -3):,}"

    # Linear data mutation: adding a clean localized string representation for values
    processed_df = df.copy()
    processed_df['formatted_salary'] = processed_df['salary'].apply(lambda x: f"€{x:,}/yr")

    return total_listings, avg_salary, processed_df

def get_svg_for_category(category_str: str) -> str:
    """Maps industry classifications to vector laboratory/chemical icons."""
    if "engineering" in str(category_str):
        # Industrial structural/gear process icon
        return '''<svg width="24" height="24" viewBox="0 0 24 24" fill="none" stroke="#74b9ff" stroke-width="2"><path d="M14.7 6.3a1 1 0 0 0 0 1.4l1.6 1.6a1 1 0 0 0 1.4 0l3.77-3.77a6 6 0 0 0-7.94 1l-1.37 1.37zM2 22l6.3-6.3M15 15l6-6M9 9l6-6M3 11a8 8 0 0 0 12.3 9.7l-1.4-1.4a6 6 0 0 1-7.9-7.9l-1.4-1.4A7.94 7.94 0 0 0 3 11z"/></svg>'''
    elif "chemistry" in str(category_str):
        # Laboratory Flask beaker vector
        return '''<svg width="24" height="24" viewBox="0 0 24 24" fill="none" stroke="#a29bfe" stroke-width="2"><path d="M6 3h12M9 3v6M15 3v6M5 21h14M5 21a2 2 0 0 1-1.9-2.6L7.3 9h9.4l4.2 9.4A2 2 0 0 1 19 21H5z"/></svg>'''
    else:
        return '''<svg width="24" height="24" viewBox="0 0 24 24" fill="none" stroke="#ffeaa7" stroke-width="2"><rect x="2" y="7" width="20" height="14" rx="2"/><path d="M16 21V5a2 2 0 0 0-2-2h-4a2 2 0 0 0-2 2v16"/></svg>'''

# ==========================================
# PHASE 3: UI RENDERING
# ==========================================
def render_job_board(search_query: str, total: int, avg_pay: str, jobs_df: pd.DataFrame):
    """Compiles the finalized engineering dataset variables directly into a modern UI card layout."""

    jobs_html = ""
    if jobs_df.empty:
        jobs_html = '<div style="text-align:center; padding: 40px; opacity: 0.6;">No chemistry openings match your scope.</div>'
    else:
        for _, row in jobs_df.iterrows():
            icon_svg = get_svg_for_category(row['category'])
            type_color = "#74b9ff" if row['type'] == "Full-Time" else "#ffeaa7" if row['type'] == "Contract" else "#55efc4"

            jobs_html += f'''
            <div style="background: rgba(255,255,255,0.04); padding: 16px; border-radius: 16px; margin-bottom: 12px; display: flex; align-items: center; justify-content: space-between; border-left: 3px solid {type_color};">
                <div style="display: flex; align-items: center; gap: 14px;">
                    <div style="background: rgba(255,255,255,0.07); padding: 10px; border-radius: 12px; display: flex; align-items: center;">
                        {icon_svg}
                    </div>
                    <div>
                        <div style="font-weight: bold; font-size: 15px; color: white;">{row['title']}</div>
                        <div style="font-size: 13px; color: rgba(255,255,255,0.6); margin-top: 2px;">{row['company']} • 📍 {row['location']}</div>
                    </div>
                </div>
                <div style="text-align: right;">
                    <div style="font-weight: 600; font-size: 14px; color: #55efc4;">{row['formatted_salary']}</div>
                    <span style="display: inline-block; font-size: 10px; font-weight: bold; padding: 3px 8px; border-radius: 20px; background: rgba(255,255,255,0.08); color: {type_color}; margin-top: 6px; text-transform: uppercase;">{row['type']}</span>
                </div>
            </div>
            '''

    display_title = f'"{search_query}" Fields' if search_query else "All Chemical Openings"

    html_layout = f'''
    <div style="max-width: 450px; margin: 20px auto; background: #0f172a; border-radius: 30px; padding: 24px; color: white; font-family: -apple-system, BlinkMacSystemFont, sans-serif; box-shadow: 0 12px 40px rgba(0,0,0,0.65);">
        <!-- Board Header Banner -->
        <div style="display: flex; justify-content: space-between; align-items: center; margin-bottom: 20px;">
            <div>
                <span style="font-size: 11px; font-weight: bold; color: #a29bfe; text-transform: uppercase; letter-spacing: 1.2px;">Science & Tech Board</span>
                <h1 style="margin: 4px 0 0 0; font-size: 24px; font-weight: 700; color: #f8fafc;">{display_title}</h1>
            </div>
            <div style="background: rgba(255,255,255,0.08); border-radius: 50%; width: 38px; height: 38px; display: flex; align-items: center; justify-content: center; font-size: 14px;">🧪</div>
        </div>

        <!-- Metrics Display Row -->
        <div style="display: flex; gap: 12px; margin-bottom: 24px;">
            <div style="flex: 1; background: linear-gradient(135deg, #1e293b, #0f172a); padding: 14px; border-radius: 18px; text-align: center; border: 1px solid rgba(255,255,255,0.05);">
                <div style="font-size: 11px; color: rgba(255,255,255,0.5); text-transform: uppercase;">Active Sectors</div>
                <div style="font-size: 22px; font-weight: 700; margin-top: 4px; color: #74b9ff;">{total}</div>
            </div>
            <div style="flex: 1; background: linear-gradient(135deg, #022c22, #0f172a); padding: 14px; border-radius: 18px; text-align: center; border: 1px solid rgba(255,255,255,0.05);">
                <div style="font-size: 11px; color: rgba(255,255,255,0.5); text-transform: uppercase;">Avg Compensation</div>
                <div style="font-size: 22px; font-weight: 700; margin-top: 4px; color: #55efc4;">{avg_pay}</div>
            </div>
        </div>

        <!-- Scrollable Feed Area -->
        <div>
            <h3 style="margin: 0 0 14px 0; font-size: 12px; color: rgba(255,255,255,0.4); letter-spacing: 0.5px; text-transform: uppercase;">Industry Openings Found</h3>
            {jobs_html}
        </div>
    </div>
    '''
    display(HTML(html_layout))

# ==========================================
# MAIN EXECUTION ROUTINE
# ==========================================
if __name__ == "__main__":
    # PROCEDURAL DATA CONTROLS: Change these directly to view targeted slices of the data pipeline.
    # Ex: set target_keyword = "Engineer" or target_location = "Stuttgart"
    target_keyword = ""
    target_location = None

    # Step 1: Linear sequence fetch pass[cite: 1]
    raw_listings = fetch_jobs(keyword=target_keyword, location=target_location)

    # Step 2: Intermediate state translation to transform data[cite: 1]
    total_count, average_pay, final_jobs_df = process_job_metrics(raw_listings)

    # Step 3: Pure programmatic execution map out to visual UI layout context[cite: 1]
    render_job_board(search_query=target_keyword, total=total_count, avg_pay=average_pay, jobs_df=final_jobs_df)
```
